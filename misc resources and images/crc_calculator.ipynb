```python
import zlib

dest_mac = bytes.fromhex("FFFFFFFFFFFF")
src_mac  = bytes.fromhex("2EFC32BA7C99")
length   = (5).to_bytes(2, "big")           
message  = bytes.fromhex("68656c6c6f")     
padding  = bytes(46 - len(message))       

data = dest_mac + src_mac + length + message + padding # bytes object

crc = zlib.crc32(data)
print(f"CRC value:      {crc:08X}")
print(f"Send order:     {crc.to_bytes(4, 'little').hex().upper()}")
```
